```python
import csv
import time
 
from google_play_scraper import Sort, reviews
 
APP_ID = "com.moveinsync.ets"   # from the Play Store URL (?id=...)
LANG = "en"
COUNTRY = "in"                  # hl=en_IN -> English, India
MAX_REVIEWS = 100              # change this to get more/fewer
OUTPUT_FILE = "moveinsync_reviews.csv"
 
 
def fetch_newest_reviews(max_reviews):
    all_reviews = []
    token = None
 
    while len(all_reviews) < max_reviews:
        batch_size = min(200, max_reviews - len(all_reviews))
        batch, token = reviews(
            APP_ID,
            lang=LANG,
            country=COUNTRY,
            sort=Sort.NEWEST,          # newest first
            count=batch_size,
            continuation_token=token,  # None on first call
        )
        all_reviews.extend(batch)
        print(f"Fetched {len(all_reviews)} reviews...")
 
        if token is None or not batch:  # no more reviews
            break
        time.sleep(1)                   # be polite to the server
 
    return all_reviews
 
 
def save_to_csv(data, filename):
    # utf-8-sig so Excel displays special characters correctly
    with open(filename, "w", newline="", encoding="utf-8-sig") as f:
        writer = csv.writer(f)
        writer.writerow(["date", "stars", "review"])
        for r in data:
            writer.writerow([
                r["at"].strftime("%Y-%m-%d %H:%M:%S"),
                r["score"],
                (r["content"] or "").replace("\n", " "),
            ])
    print(f"Saved {len(data)} reviews to {filename}")
 
 
if __name__ == "__main__":
    data = fetch_newest_reviews(MAX_REVIEWS)
    save_to_csv(data, OUTPUT_FILE)
```

```python
import csv
import time
 
from google_play_scraper import Sort, reviews
 
APP_ID = "com.moveinsync.ets"   # from the Play Store URL (?id=...)
LANG = "en"
COUNTRY = "in"                  # hl=en_IN -> English, India
MAX_REVIEWS = 500             # change this to get more/fewer
OUTPUT_FILE = "moveinsync_reviews_new.csv"
 
 
def fetch_newest_reviews(max_reviews):
    all_reviews = []
    token = None
 
    while len(all_reviews) < max_reviews:
        batch_size = min(200, max_reviews - len(all_reviews))
        batch, token = reviews(
            APP_ID,
            lang=LANG,
            country=COUNTRY,
            sort=Sort.NEWEST,          # newest first
            count=batch_size,
            continuation_token=token,  # None on first call
        )
        all_reviews.extend(batch)
        print(f"Fetched {len(all_reviews)} reviews...")
 
        if token is None or not batch:  # no more reviews
            break
        time.sleep(1)                   # be polite to the server
 
    return all_reviews
 
 
def save_to_csv(data, filename):
    # utf-8-sig so Excel displays special characters correctly
    with open(filename, "w", newline="", encoding="utf-8-sig") as f:
        writer = csv.writer(f)
        writer.writerow(["date", "stars", "review"])
        for r in data:
            writer.writerow([
                r["at"].strftime("%Y-%m-%d %H:%M:%S"),
                r["score"],
                (r["content"] or "").replace("\n", " "),
            ])
    print(f"Saved {len(data)} reviews to {filename}")
 
 
if __name__ == "__main__":
    data = fetch_newest_reviews(MAX_REVIEWS)
    save_to_csv(data, OUTPUT_FILE)
```

```python
import pandas as pd, matplotlib.pyplot as plt
df = pd.read_csv("moveinsync_reviews_new.csv", parse_dates=["date"])
df["week"] = df["date"].dt.to_period("W").dt.start_time
wk = df.groupby("week").agg(
    n=("stars","size"),
    low=("stars", lambda s: (s<=2).mean()*100)
)
fig, ax = plt.subplots(figsize=(10,4))
ax.plot(wk.index, wk["low"], marker="o")
ax.axvspan(pd.Timestamp("2026-09-21"), pd.Timestamp("2026-10-04"), alpha=.15, color="red")
ax.set_ylabel("Low-star share (%)"); ax.set_title("Weekly low-star share")
plt.tight_layout(); plt.savefig("trend.png", dpi=150)
```
